```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("/content/rta_enterprise_dataset.csv")
df.head()
```

```python
## Problem Statement

##Business Objective:

##Increase the average Service Level from 73% to at least 80% while maintaining customer satisfaction and operational efficiency.

## Potential Root Causes

##Based on the exploratory analysis, several operational factors may contribute to the low Service Level:

##- High Average Speed of Answer (ASA)
##- High Occupancy
##- Increased Call Volume
##- Customer Abandonment
##- Forecasting inaccuracies
```

```python
numeric = df.select_dtypes(include="number")

corr = numeric.corr()
root_causes = pd.DataFrame({

"Factor":[
"Average Speed of Answer",
"Occupancy",
"Abandon Rate",
"Actual Calls",
"Forecast Calls",
"AHT",
"Adherence"
],

"Correlation_with_SLA":[
corr.loc["ASA","SLA"],
corr.loc["Occupancy","SLA"],
corr.loc["Abandon_Rate","SLA"],
corr.loc["Actual_Calls","SLA"],
corr.loc["Forecast_Calls","SLA"],
corr.loc["AHT","SLA"],
corr.loc["Adherence","SLA"]
]

})

root_causes
```

```python
root_causes = root_causes.sort_values(
    by="Correlation_with_SLA",
    key=abs,
    ascending=False
)

root_causes
```

## Conclusion

The analysis suggests that the primary drivers of low Service Level are:

- High customer waiting time (ASA)
- High agent Occupancy
- High customer Abandon Rate

Improving these operational areas is expected to have the greatest impact on Service Level performance.
