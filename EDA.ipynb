## Correlation Analysis

The correlation analysis identified several key operational relationships.

- Average Speed of Answer (ASA) has the strongest negative relationship with Service Level (-0.999), indicating that longer customer waiting times significantly reduce SLA performance.

- Occupancy also demonstrates a strong negative correlation (-0.889), suggesting that highly utilized agents struggle to maintain target response times.

- Customer Satisfaction shows a nearly perfect positive relationship with SLA (0.999), highlighting the importance of operational responsiveness for customer experience.

- Call volume exhibits a moderate negative relationship with SLA, indicating that increased demand places additional pressure on operational performance.

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Correlation Matrix of Operational KPIs")

plt.show()
```

## Business Interpretation

The correlation matrix reveals several important operational relationships.

- Service Level decreases as customer waiting time (ASA) increases.
- Higher Occupancy is associated with lower SLA and higher customer abandonment.
- Increased call volume contributes to higher Occupancy and lower Service Level.
- Workforce planning appears to scale staffing with demand, as Forecasted Calls, Actual Calls, and Scheduled Agents are strongly correlated.
- Average Handle Time (AHT) shows little relationship with the remaining KPIs, suggesting that waiting time and staffing are the primary operational drivers in this dataset.

These findings provide direction for the Root Cause Analysis performed in the next stage of the project.

```python
sla_queue = (
    df.groupby("Queue")["SLA"]
      .mean()
      .sort_values()
)

print(sla_queue)

plt.figure(figsize=(8,5))

sla_queue.plot(kind="bar")

plt.title("Average Service Level by Queue")
plt.xlabel("Queue")
plt.ylabel("Average SLA (%)")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
calls_queue = df.groupby("Queue")["Actual_Calls"].mean().sort_values()

plt.figure(figsize=(8,5))

calls_queue.plot(kind="bar")

plt.title("Average Calls by Queue")

plt.ylabel("Average Calls")

plt.xticks(rotation=45)

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
daily_calls = df.groupby("DayOfWeek")["Actual_Calls"].sum()

daily_calls = daily_calls.reindex([
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
])

plt.figure(figsize=(9,5))

daily_calls.plot(kind="bar")

plt.title("Total Calls by Day of Week")

plt.ylabel("Calls")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
df["Hour"] = df["Interval"].str[:2].astype(int)
```

```python
hourly = df.groupby("Hour")["Actual_Calls"].mean()

plt.figure(figsize=(10,5))

hourly.plot()

plt.title("Average Calls by Hour")

plt.xlabel("Hour")

plt.ylabel("Average Calls")

plt.grid(True)

plt.show()
```

```python
sla_hour = df.groupby("Hour")["SLA"].mean()

plt.figure(figsize=(10,5))

sla_hour.plot()

plt.title("Average SLA by Hour")

plt.xlabel("Hour")

plt.ylabel("SLA (%)")

plt.grid(True)

plt.show()
```
