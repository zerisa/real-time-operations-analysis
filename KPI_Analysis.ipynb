# KPI Performance Analysis

## Objective

The objective of this notebook is to evaluate the operational performance of the contact center by analyzing its key performance indicators (KPIs).

The analysis focuses on identifying operational strengths, potential bottlenecks, and opportunities for improvement.

```python
## KPI 1 - Service Level (SLA)

## Service Level measures the percentage of customer contacts answered within the target response time.

## Target SLA: **80%**
```

```python

import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

df = pd.read_csv("/content/rta_enterprise_dataset.csv")
df.head()
```

```python
average_sla = df["SLA"].mean()

print(f"Average SLA: {average_sla:.2f}%")
```

```python
target_sla = 80

gap = average_sla - target_sla

print(f"Gap to Target: {gap:.2f}%")

### Interpretation

#The average Service Level is below the organizational target of 80%.

#This indicates that customers experience longer waiting times than expected during certain operational periods.
```

```python
## KPI 2 - Occupancy

## Occupancy measures how much of an agent's available time is spent handling customer interactions.

## Recommended operational range: **80%–85%**

avg_occ = df["Occupancy"].mean()

print(avg_occ)
```

```python
target_occ = 85

gap_occ = avg_occ - target_occ

print(gap_occ)

#below target
```

```python
avg_asa = df["ASA"].mean()

print(avg_asa)
```

```python
target_asa = 20
print(avg_asa-target_asa)

#Average customer waiting time exceeds the desired operational target.
#Reducing ASA is expected to improve both Service Level and Customer Satisfaction.
```

```python
avg_csat = df["CSAT"].mean()

print(avg_csat)
```

```python
target_csat = 4.5

#Customer Satisfaction remains below the desired target, indicating opportunities to improve the customer experience.
```

```python
import numpy as np

df["Forecast_Accuracy"] = np.where(
    df["Actual_Calls"] > 0,
    1 - abs(df["Forecast_Calls"] - df["Actual_Calls"]) / df["Actual_Calls"],
    np.nan
)
```

```python
df["Forecast_Accuracy"].mean()
```

```python
summary = {
    "KPI": [
        "SLA",
        "Occupancy",
        "ASA",
        "CSAT",
        "Forecast Accuracy"
    ],
    "Current": [
        df["SLA"].mean(),
        df["Occupancy"].mean(),
        df["ASA"].mean(),
        df["CSAT"].mean(),
        df["Forecast_Accuracy"].mean()
    ],
    "Target": [
        80,
        85,
        20,
        4.5,
        0.90
    ]
}

summary_df = pd.DataFrame(summary)

summary_df
```
